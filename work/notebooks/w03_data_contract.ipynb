## 1. Unit of analysis + time window

*One row = one what, over which dates? State it, then verify it below.*

**One row = one content page for one client, summarized over March 2026.** The daily warehouse table is at report-date × client × content grain, so I aggregate it to one client × content row for the March 2026 snapshot. The prediction/ranking decision uses information available by the end of March 2026, while the decline label is treated as the outcome to evaluate separately.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# March 2026 snapshot: verify the raw warehouse grain

import duckdb

con = duckdb.connect()

query = """
SELECT
    COUNT(*) AS rows,
    COUNT(DISTINCT client_id) AS clients,
    COUNT(DISTINCT content_id) AS contents,
    MIN(report_date) AS min_date,
    MAX(report_date) AS max_date
FROM read_parquet(
    'hf://datasets/FlyRank/internship-warehouse/fact_content_daily_performance/month=2026-03/*.parquet'
)
"""

con.sql(query)
```

## 3. Verify it with queries (grain, counts, missing values, windows)

*Every claim above gets a query cell here. A contract claim without a query next to it is a guess.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

# Load the same lane data used in Week 2
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Create the label
df["is_declining_label"] = (
    df["trend_direction"].eq("down")
).astype(int)

# ============================================================
# QUERY 1 — GRAIN
# ============================================================

grain_check = (
    df.groupby("content_id")
      .size()
      .reset_index(name="rows_per_content")
)

print("QUERY 1 — Grain")
print("Unique content pages:", grain_check["content_id"].nunique())
print("Maximum rows per content:", grain_check["rows_per_content"].max())
print("Rows with duplicates:", (grain_check["rows_per_content"] > 1).sum())


# ============================================================
# QUERY 2 — ROW COUNT + DATE WINDOW
# ============================================================

print("\nQUERY 2 — Row count and date span")

if "report_date" in df.columns:
    df["report_date"] = pd.to_datetime(df["report_date"])

    march = df[
        (df["report_date"] >= "2026-03-01") &
        (df["report_date"] < "2026-04-01")
    ]

    print("March 2026 rows:", len(march))
    print("Minimum date:", march["report_date"].min())
    print("Maximum date:", march["report_date"].max())
else:
    # The anonymized Week-2 file has no report_date.
    # Treat the file itself as the March development slice.
    march = df.copy()

    print("March 2026 rows:", len(march))
    print("Date column: not available in anonymized file")


# ============================================================
# QUERY 3 — AVAILABILITY
# IS TRUE equivalent: explicitly keep available/valid rows
# ============================================================

print("\nQUERY 3 — Availability")

availability_col = None

for col in ["ga4_data_available", "available", "is_available"]:
    if col in march.columns:
        availability_col = col
        break

if availability_col:
    available = march[march[availability_col].eq(True)]
    print("Availability field:", availability_col)
    print("Rows surviving IS TRUE:", len(available))
else:
    # No GA4 availability column in the anonymized Week-2 file.
    # Keep rows with the required historical signals present.
    available = march[
        march["impressions_90d"].notna()
        & march["avg_position"].notna()
        & march["ctr"].notna()
    ]

    print("Availability proxy: required historical fields present")
    print("Rows surviving availability check:", len(available))
```
