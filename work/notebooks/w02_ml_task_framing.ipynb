## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule could rank pages using simple thresholds such as page age or recent traffic change, but several signals can interact. Content age, days since update, impressions, position, CTR, and word count may each provide useful information. ML can learn combinations of these signals instead of requiring every condition to be written as a separate if-statement. The ranking output can support a review queue by putting higher-priority pages first.
