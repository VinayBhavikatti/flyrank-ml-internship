## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

url = "https://raw.githubusercontent.com/VinayBhavikatti/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Rows:", len(df))
print("Columns:", list(df.columns))

print("\nUseful numbers for my lane:")
print("Total pages:", len(df))
print("Average impressions (90d):", round(df["impressions_90d"].mean(), 2))
print("Average clicks (90d):", round(df["clicks_90d"].mean(), 2))
print("Average position:", round(df["avg_position"].mean(), 2))
```

The starter dataset contains 30,000 content pages. On average, pages received 5200.37 impressions and 16.11 clicks over 90 days, with an average position of 16.34. These numbers show that the dataset contains measurable search performance differences that can be used to identify which pages may deserve review first.
